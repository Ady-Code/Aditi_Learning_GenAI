## Scenario: E-commerce Order Processing System

You run an online store.
You have orders. You want to:

* Filter valid orders
* Apply discounts
* Calculate totals

We’ll use:

* `lambda`
* `map`
* `filter`
* `reduce`

```python
# Step 0 — Sample Data

orders = [
    {"id": 1, "customer": "Amit", "amount": 1200, "status": "completed"},
    {"id": 2, "customer": "Neha", "amount": 800, "status": "cancelled"},
    {"id": 3, "customer": "Raj", "amount": 1500, "status": "completed"},
    {"id": 4, "customer": "Sara", "amount": 400, "status": "completed"},
]
```

You need to perform below : 
- Filter all completed orders using python fucntion.
- Filter all completed orders using lambda fucntion.
- Apply 10% discount on all completed orders using python fucntion.
- Apply 10% discount on all completed orders using lambda fucntion.
- Calcualte total revenue by adding all amounts of all completed orders using python function. 
- Calcualte total revenue by adding all amounts of all completed orders using lambda function. 
- Apply user defined discount whcih can change on all completed orders. (hint you might need to use lambda function here along with python function)

```python
# Code for filter all completed orders using python fucntion 

orders = [
    {"id": 1, "customer": "Amit", "amount": 1200, "status": "completed"},
    {"id": 2, "customer": "Neha", "amount": 800, "status": "cancelled"},
    {"id": 3, "customer": "Raj", "amount": 1500, "status": "completed"},
    {"id": 4, "customer": "Sara", "amount": 400, "status": "completed"},
]
result = []
for order in orders:
    if order["status"]=='completed':
        result.append(order)

print(result)
```

```python
# Code for filter all completed orders using lambda fucntion 
orders = [
    {"id": 1, "customer": "Amit", "amount": 1200, "status": "completed"},
    {"id": 2, "customer": "Neha", "amount": 800, "status": "cancelled"},
    {"id": 3, "customer": "Raj", "amount": 1500, "status": "completed"},
    {"id": 4, "customer": "Sara", "amount": 400, "status": "completed"},
]
completed_order = list(filter(lambda order: order["status"]=='completed', orders))
print(completed_order)
```

```python
# Code to apply 10% discount on all completed orders using python fucntion 

orders = [
    {"id": 1, "customer": "Amit", "amount": 1200, "status": "completed"},
    {"id": 2, "customer": "Neha", "amount": 800, "status": "cancelled"},
    {"id": 3, "customer": "Raj", "amount": 1500, "status": "completed"},
    {"id": 4, "customer": "Sara", "amount": 400, "status": "completed"},
]

result = []
def disc_order():
    for order in orders:
        if order["status"] == "completed":
            result.append(order)
            order["amount"] = order["amount"] * 0.9
    print(result)

disc_order()
```

```python
# Code to apply 10% discount on all completed orders using lambda fucntion 

orders = [
    {"id": 1, "customer": "Amit", "amount": 1200, "status": "completed"},
    {"id": 2, "customer": "Neha", "amount": 800, "status": "cancelled"},
    {"id": 3, "customer": "Raj", "amount": 1500, "status": "completed"},
    {"id": 4, "customer": "Sara", "amount": 400, "status": "completed"},
]

completed_order = list(filter(lambda order: order["status"]=='completed', orders))
# disc_order = list(map(lambda comp_order:comp_order["amount"] * 0.9, completed_order))

def disc_order(order):
    order["amount"] = order["amount"] * 0.9
    return order

# disc_order = list(map(lambda order : order["amount"]*0.9 , orders))
result = list(map(disc_order ,completed_order))
print(result)
```

```python
# Apply user defined discount whcih can change on all completed orders using python function.

orders = [
    {"id": 1, "customer": "Amit", "amount": 1200, "status": "completed"},
    {"id": 2, "customer": "Neha", "amount": 800, "status": "cancelled"},
    {"id": 3, "customer": "Raj", "amount": 1500, "status": "completed"},
    {"id": 4, "customer": "Sara", "amount": 400, "status": "completed"},
]

def apply_discount():
    discount = 20
    for order in orders:
        if order["status"] == "completed":
            # print(order)
            # final_result = list(filter(lambda order:order["amount"] * (1-discount/100), orders))
            # print(final_result)
            order["amount"] = order["amount"] * (1-discount/100)
            print(order)

apply_discount()
```
