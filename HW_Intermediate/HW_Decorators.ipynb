```python
# Level 2: The Stopwatch Decorator ⏱️
# Import time.
# Create a decorator @timer.
# Inside the wrapper, record the start time (start = time.time()).
# Run the original function.
# # Record the end time (end = time.time()).
# Print "Execution time: [X] seconds".
# Apply it to a function that loops 1 million times.

import time
def timer(func):
    def time_wrapper():
        start = time.time()
        print("********************") # Do something BEFORE
        func()                      # Run the original function
        print("********************")
        end = time.time()
        print(f"Execution time: {end - start} seconds")
    return time_wrapper

@timer
def loop_func():
    for i in range(100000):
        pass

loop_func()
```

```python
# Level 3: The Slow Down Decorator 🐢
# Import time.
# Create a decorator @slow_down.
# Inside the wrapper, make the program sleep for 2 seconds (time.sleep(2)) before running the original function.
# Apply it to a countdown(n) function.

import time
def slow_down(func):
    def sleep_mode():
        time.sleep(2)
    return sleep_mode

@slow_down
def countdown(n):
    print(f"countdown: {n}")
    n -= 1

countdown()
```

```python
# Level 4: The Debugger Decorator 🐛
# Create @debug.
# The wrapper should print exactly what arguments the function was called with. (Hint: Print args and kwargs inside the wrapper).
# Run the function, save the result.
# Print the result.
# Return the result.
# Apply it to multiply(a, b).

def debug(func):
    def wrapper(*args, **kwargs):
        print(f"Running the function: {func.__name__} with {args} and {kwargs}")
        result = func(*args,**kwargs)
        print("✅ Finished!")
        return result
    return wrapper

@debug
def multiply(a: int, b: int) -> int:
    return a * b

print("Result:", multiply(5, 10))
```

```python
# Level 5: The Resilient Server (@retry) (Real Scenario) 📡
# Scenario: You have a function that connects to a database. Sometimes the internet drops, and it crashes. You want a decorator that automatically tries 3 times before finally crashing.

# Create a decorator @retry.
# Inside the wrapper, use a for loop that runs 3 times.
# Inside the loop, use try/except.
# Try to run the function and return the result.
# If it crashes (Exception), print "Failed, retrying...".
# If the loop finishes all 3 tries and still fails, raise the Exception.
# Apply it to the provided connect_to_db function.

import random

def retry(func):
    def wrapper(*args,**kwargs):
        for i in range(3):
            try:
                return func(*args,**kwargs)
            except Exception as e:
                print("Failed, retrying...")
            if i == 2:
                raise e
    return wrapper

@retry
def connect_to_db():
    if random.choice([True, False, False]): # 66% chance to crash!
        raise ConnectionError("Database timeout!")
    return "Connected successfully!"

print(connect_to_db())
```
