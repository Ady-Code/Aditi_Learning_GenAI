```python
# Level 1: Time Travel (timedelta)
# Import datetime and timedelta.
# Get the current naive local time.
# Calculate the date exactly 30 days and 12 hours from right now.
# Print the future date.

from datetime import datetime as dt, timedelta as td

date_now = dt.now()
print(date_now)

date_30d = date_now + td(days=30,hours=12)
print(date_30d)
```

```python
# Level 2: Zone Aware (zoneinfo / pytz)
# Import ZoneInfo from zoneinfo (or timezone from pytz).
# Create a timezone-aware datetime object for the current time in "Asia/Tokyo".
# Create another timezone-aware datetime for "America/New_York".
# Print both to see the time difference!

from zoneinfo import ZoneInfo as zi 

time_tokyo = dt.now(zi("Asia/Tokyo"))
time_NY = dt.now(zi("America/New_York"))
print(time_tokyo)
print(time_NY)
```

```python
# Level 3: Modern Paths (pathlib) 📁
# Import Path from pathlib.
# Create a Path object for a new directory called "project_files".
# Use .mkdir(exist_ok=True) to create it.
# Inside that directory, create a new Path object for "log.txt" and use .touch() to create the empty file.
# Print whether the file .exists().

from pathlib import Path

parent_path = Path("project_files")
parent_path.mkdir(parents=True, exist_ok=True)
file_path = parent_path / "log.txt"
file_path.touch() # creates empty file
print(file_path.exists())
```

```python
# Level 4: The Mover (shutil + pathlib) 🚚
# Import shutil and Path.
# Create a Path for a new folder called "backup". Make the folder.
# Use shutil.copy() to copy the "log.txt" (from Level 3) into the "backup" folder.
# Check if the backup file exists using Pathlib.

import shutil
parent_path = Path("backup").mkdir(parents=True, exist_ok=True)
shutil.copy(Path("project_files") / "log.txt",Path("backup") / "log1.txt")
new_file = Path("backup") / "log1.txt"
print(new_file.exists())
```

```python
# Level 5: The "Old File" Cleaner (Advanced) 👴
# Scenario: Clean up old logs.
# Define a "cutoff date" (e.g., 30 days ago).
# Compare it to a fake file date.
# Logic:
# file_date = datetime(2020, 1, 1) (Very old)
# if file_date < cutoff_date: print("Delete this file")
# else: print("Keep this file")

from datetime import datetime, timedelta

cutoff_date = datetime.now() - timedelta(days=30)
print(cutoff_date)
file_date = datetime(2020, 1, 1)
print(file_date)
if file_date < cutoff_date: 
    print("Delete this file")
else: print("Keep this file")
```
