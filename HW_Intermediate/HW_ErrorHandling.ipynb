```python
# Level 1: The Safe Calculator (Else) 🧮
# Write a function safe_square(n).
# try to convert n to an integer and square it.
# except a ValueError (print "Not a number").
# Task: Use the else block to print "The square is [X]".
# Test with "5" and "hello".

def safe_square(n):
    try:
        result = int(n) * int(n)
    except ValueError:
        print("Not a number")
    else:
        print(f"The square is {result}")
    finally:
        print("Successfully reached the end")

safe_square("5")
safe_square("hello")
```

```python
# Level 2: The File Closer (Finally) 📁
# Scenario: You are opening a file. Even if it crashes, you must close it.

# Create a dummy file temp.txt.
# Write a try block that opens it.
# Inside try, raise a ZeroDivisionError (simulate a crash).
# Use finally to close the file and print "File Closed Safely".
import os
# os.getcwd()
try:
    dummy = open(os.path.join("project_files","temp.txt"), "w")
except ZeroDivisionError:
    print("Exiting the try-catch block")
else:
    print("File opened successfully")
finally:
    dummy.close()
    print("File Closed Safely")

```

```python
# Level 3: The Custom Password Error 🔐
# Create a custom exception WeakPasswordError.
# Write a function register(password).
# If password length < 6, raise WeakPasswordError("Too short!").
# Wrap the call in try/except to catch that specific error.

class WeakPasswordError(Exception):
    print("Custom Error 'WeakPasswordError' defined successfully.")

def register(password):
    if len(password) < 6:
        raise WeakPasswordError()
    
try:
    register("abc")
except WeakPasswordError as e:
    print("Too short!")
```

```python
# Level 4: The Login System (Multiple Custom Errors) 🛑
# Create two custom exceptions: UserNotFoundError and WrongPasswordError. Task:

# Dictionary: db = {"admin": "1234"}.
# Function login(user, pwd).
# Check if user exists 
#  raise UserNotFoundError.
# Check if pwd matches 
#  raise WrongPasswordError.
# Test with wrong user AND wrong password.

class UserNotFoundError(Exception):
    pass
class WrongPasswordError(Exception):
    pass
db = {"admin": "1234"}
def login(user, pwd):
    if user not in db:
        raise UserNotFoundError("User not found")
    if db[user] != pwd:
        raise WrongPasswordError("Wrong pwd")
        
try:
    login("admin", "223")
except UserNotFoundError as e: # learnt 2 except in one try block
    print(e)
except WrongPasswordError as e1:
    print(e1)
else:
    print("Login successful")
finally:
    print("Exiting the function")
```

```python
# Level 5: The Reliable Server (Complex Logic) 📡
# Scenario: Simulate a connection retry system.

# Create a custom ConnectionError.
# Create a function connect(attempt).
# Logic: If attempt < 3, raise ConnectionError. If attempt == 3, return "Success".
# Loop: Try connecting 5 times using a for loop.
# Inside loop: try/except/else/finally.
# Catch ConnectionError: Print "Retrying...".
# else: Print "Connected!" and break.
# finally: Print "Attempt finished".

class ConnectionError(Exception):
    pass

def connect(attempt):
    if attempt < 3:
        raise ConnectionError("Retrying...")
    else:
        return "Success"

for i in range(1,6): #learnt putting try block in a loop while calling the function
    try:
        connect(i)
    except ConnectionError as e:
        print(e)
    else:
        print("Connected!")
    finally:
        print("Attempt finished")
```
