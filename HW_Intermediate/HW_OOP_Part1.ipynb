```python
# Level 1: The Basic Blueprint 🏗️
# Create a class Book.
# __init__ should take title and author.
# Create an Instance Method called read() that prints: "You are reading [title] by [author]".
# Create an object and call .read().
class Book:
    def __init__(self,title,author):
        self.title = title
        self.author = author
    def read(self):
        print(f"You are reading {self.title} by {self.author}")

read_book = Book("Power","Robert Greene")
read_book.read()
```

```python
# Level 2: State Tracking 🎛️
# Create a class SmartBulb.
# __init__ should set self.is_on = False.
# Create a method toggle(): If it's on, turn it off. If it's off, turn it on.
# Print the state after each toggle.
class SmartBulb:
    def __init__(self):
        self.is_on = False
    
    def toggle(self):
        if self.is_on:
            self.is_on = False
        else:
            self.is_on = True
        print("Bulb is ON" if self.is_on else "Bulb is OFF")
    
bulb = SmartBulb()
bulb.toggle()
bulb.toggle()
bulb.toggle()
```

```python
# Level 3: The Secret Vault (Encapsulation) 🔐
# Create a class Vault.
# __init__ should take a password and store it as a private variable (__password).
# Write a standard instance method called check_password(self, guess).
# Return True if guess matches the private password, else False.
class vault:
    def __init__(self,password):
        self.__password = password
    def check_password(self, guess):
        if guess == self.__password:
            return True
        else:
            return False
check = vault("secret111")
print(check.check_password("secret111"))
print(check.check_password("wrong"))
```

```python
# Level 4: The Thermostat (@property) 🌡️
# Create a class Thermostat.
# __init__ sets a private variable __temperature.
# Create a Getter using @property for temperature.
# Create a Setter using @temperature.setter.
# Rule: If the new temperature is over 100, print "Too hot!" and don't change it. Otherwise, update the private variable.
class Thermostat:
    def __init__(self,temperature):
        self.__temperature = temperature
    
    @property
    def temperature(self):
        return self.__temperature
    @temperature.setter
    def temperature(self, new_temperature):
        if new_temperature > 100:
            print("Too hot!")
        else:
            self.__temperature = new_temperature

temp = Thermostat(22)
print(temp.temperature)
temp.temperature = 150 
print(temp.temperature)
```

```python
# Level 5: The Secure Bank (Real Scenario) 🏦
# Scenario: Build a Bank Account that completely protects its money.

# Class BankAccount.
# __init__(owner, initial_balance): Sets public owner and private __balance.
# @property balance: Getter to view the balance.
# Method deposit(amount): If amount > 0, add to __balance. Else, print error.
# Method withdraw(amount): If amount > __balance, print "Insufficient Funds". Else, subtract it. Notice: We do NOT use a @balance.setter here, because we only want balance changed via official deposit/withdraw methods!

class BankAccount:
    def __init__(self,owner, initial_balance):
        self.owner = owner
        self.__balance = initial_balance
    
    @property
    def balance(self):
        return self.__balance
    
    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
        else:
            raise ValueError("Amount cannot be negative!")

    def withdraw(self,amount):
        if amount > self.__balance:
            print("Insufficient Funds")
        else:
            self.__balance -= amount

account = BankAccount("Alice",5000)   
account.deposit(1000)     
print(account.balance)
account.withdraw(2000)
print(account.balance)
```
