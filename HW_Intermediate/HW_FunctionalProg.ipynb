```python
# Level 1: The Lambda Square 🔲
# Create a lambda function called get_square that takes n and returns n * n.
# Test it with the number 8.

get_square = lambda n : n * n
print(get_square(8))
```

```python
# Level 2: The Name Filter 🧹
# You have a list: names = ["Alice", "Bob", "Amanda", "Charlie", "Anna"].

# Use filter() and a lambda to create a list of names that start with "A".
# Print the result.

names = ["Alice", "Bob", "Amanda", "Charlie", "Anna"]

result = list(filter(lambda name: name.startswith("A"),names))
print(result)
```

```python
# Level 3: The Temperature Converter (Map) 🌡️

# Use map() to convert them to Fahrenheit.
# Formula: (C * 9/5) + 32.
# Print the new list.
celsius = [0, 20, 30, 100]
new_list = list(map(lambda c:(c * 9/5) + 32,celsius))
print(new_list)
```

```python
# Level 5: The HTML Tag Generator (**kwargs) 🏷️
# Create a function make_tag(text, **attributes).

# It should return an HTML string.
# Example Call: make_tag("Click Me", href="google.com", style="color:red")
# Expected Output: <a href="google.com" style="color:red">Click Me</a> (Hint: Loop through attributes.items() to build the string).

def make_tag(text, **attributes):
    att_str = ''
    for i,j in attributes.items():
        att_str += f' {i}="{j}"'
    return f"<a {att_str}>{text}</a>"

print(make_tag("Click Me", href="google.com", style="color:red"))
```

```python
# Level 4: The Aggregator (reduce) 🗜️
# You have a list of numbers: numbers = [1, 2, 3, 4, 5].
# Use reduce() and a lambda to multiply all the numbers together (1 * 2 * 3 * 4 * 5).
# Print the final product.

from functools import reduce
numbers = [1, 2, 3, 4, 5]
print(reduce(lambda x , y : x * y, numbers))
```
