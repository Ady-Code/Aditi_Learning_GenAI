```python
# 2. WITH Type Hinting (Professional)
def calculate_total_pro(price: float, tax: float) -> float:
    """Calculates the total price including tax."""
    return price + (price * tax)

print(calculate_total_pro(100.0, 0.05))
```

```python
# Complex types require the 'typing' module in older Python versions, 
# but in Python 3.9+, you can just use list, dict, etc.
def process_users(users: list[str]) -> dict[str, bool]:
    return {user: True for user in users}

print(process_users(["Alice", "Bob"]))
```

```python
# Rewrite this:
def repeat_word(word: str, times:int) -> str:
    return word * times

repeat_word("Python", 3)
```

```python
# 1. Assigning a function to a variable (NO parentheses!)
def lower_name(name:str) -> str:
    return name.lower()

all_lower = lower_name
print(all_lower("PYTHON"))
```

```python
# 2. Calling a function inside a function
def lower_char(name:str) -> str:
    return lower_name(name).upper()

lower_char("hello!")
```

```python
# 3. Passing a function into another function
def greet_in_lower(lower_char,name: str) -> str:
    greeting = lower_char(name)
    print(greeting)

greet_in_lower(lower_char,"aditi")
```

```python
def parent():
    print("Printing from parent()")

    def first_child():
        print("Printing from first_child()")

    def second_child():
        print("Printing from second_child()")

    second_child()
    first_child()

parent()
```
