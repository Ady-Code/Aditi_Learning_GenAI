```python
# Problem4: Nested Function Challenge

numbers = []

def process_numbers(numbers):
    def even_numbers():
        even_num = [i for i in numbers if i % 2 == 0]
        print(f"Evens: {even_num}")
    def avg_numbers():
        try:
            print(f"Avg: {sum(numbers)/len(numbers)}")

        except ZeroDivisionError:
            print("Error occurred: division by zero")
    
    even_numbers()
    avg_numbers()
    
process_numbers([1,2,4,6,10,10,20])
process_numbers([])

    
```

```python
# Problem 9: Server Metrics Dataset

server_metrics = [
    {"server": "prod-1", "cpu": 75, "memory": 68, "status": "healthy"},
    {"server": "prod-2", "cpu": 95, "memory": 88, "status": "critical"},
    {"server": "prod-3", "cpu": 45, "memory": 50, "status": "healthy"},
    {"server": "prod-4", "cpu": "N/A", "memory": 70, "status": "warning"},
    {"server": "prod-5", "cpu": 85, "memory": 92, "status": "critical"},
]

def metrics():
    # print(server_metrics[0]["cpu"])
    # server = {i["server"] for i in server_metrics if i["cpu"] > 80}
    # print(server)
    total_cpu = 0
    list_server = []
    max_usage = 0
    
    for i in server_metrics:
        if max_usage < i["memory"]:
            max_usage = i["memory"]
            memory = i["server"]
        try:
            total_cpu += i["cpu"]
            # print(total_cpu)
            if i["cpu"] > 80:
                # print(i["server"])
                list_server.append(i["server"])                
        except TypeError:
            print(f"Invalid CPU value: {i["server"]}")

    print(f"Overloaded servers (CPU > 80): {list_server}")
    print(f"Average CPU: {total_cpu/len(server_metrics)}")
    # Group by status - NOT SURE HOW TO DISPLAY THE OUTPUT
    print(f"Highest memory usage is for {memory}: {max_usage}")

metrics()
```

```python
# Problem 8: Order Processing Dataset

order_details = [
    {"order_id": 1001, "customer": "Alice", "items": [("Laptop", 1, 70000)], "status": "pending"},
    {"order_id": 1002, "customer": "Bob", "items": [("Mouse", 2, 500), ("Keyboard", 1, 1500)], "status": "shipped"},
    {"order_id": 1003, "customer": "Charlie", "items": [("Monitor", 2, 12000)], "status": "pending"},
    {"order_id": 1004, "customer": "David", "items": [], "status": "cancelled"},
    {"order_id": 1002, "customer": "Bob", "items": [("Mouse", 2, 500)], "status": "pending"},  # duplicate ID
    {"order_id": 1005, "customer": "Eva", "items": [("Laptop", 1, "invalid_price")], "status": "pending"},
]

def revenue():
    total_revenue = 0
    invalid_values = []
    order_id = []
    pending_orders = []
# traverse thru the list of dictionaries
    for orders in order_details:
            # if len(orders["items"]) >= 1:
# Skip empty item orders
            # if len(orders["items"]) == 0:
            #     print(f"Skip the item: {orders['order_id']}")
# for multiple items, traverse thru the list of tuples
            for i in range(len(orders["items"])):
                # print(orders["items"][i])
                # print(orders["items"][i][2])
# get item cost for each order, ignore the string values
                if isinstance(orders["items"][i][2],(int,float)):
                     total_revenue += orders["items"][i][2]
# Handle invalid price values
                # if isinstance(orders["items"][i][2],(str)):
                else:
                     invalid_values.append(orders["items"][i][2])
                     order_id.append(orders["order_id"])
# Process only “pending” orders
                if orders["status"] == "pending":
                     pending_orders.append(orders["order_id"])
    print(f"Total revenue: {total_revenue}")
    print(f"Invalid value: {invalid_values} found against order_id: {order_id}")
    print(f"Process only these pending orders: {pending_orders}")


# Detect duplicate order IDs
def duplicate_orders():
    dup_list = []
    dup_items = []
    for orders in order_details:
        dup_list.append(orders["order_id"])
# traverse thru the list and see if duplicates are found
    for items in set(dup_list):
        if dup_list.count(items) > 1:
             dup_items.append(items)
    print(f"Duplicate order ids: {dup_items}")
  
    # print(dup_list)
    # print(list(set(dup_list)))     

revenue()
duplicate_orders()
```
