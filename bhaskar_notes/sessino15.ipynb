nums = [2,3,4,5]

Using reduce(), calculate the product of numbers:

````python
from functools import reduce
nums = [2,3,4,5]
product_nums = reduce(lambda x,y:x*y,nums)
print(product_nums)
````

Write a function get_integer(value) that converts a value to int. If the conversion fails, catch the ValueError and return the string 'Invalid input'.


````python
def get_integer(value):
    try:
        int_num = int(value)
        print(int_num)
    except ValueError:
        print("Invalid input")

get_integer("ewqd")
get_integer("23")
````
