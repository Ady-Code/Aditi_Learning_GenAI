## Working with dates (March 05)
### `datetime` Module
- `datetime` module is used to work with dates, times, timestamps, deadlines, and durations. 
- Commonly used in real-world applications like logging, scheduling jobs, measuring execution time, and handling time zones.
- Python provides the datetime module to work with time-related data.
`import datetime`
- Main classes inside the module: 

| Class | Purpose |
|---|---|
| date | Only date (year, month, day) |
| time | Only time (hour, minute, second) |
| datetime | Date + time together |
| timedelta | Difference between two dates/times |
| timezone | Time zone handling |

- Getting current date & time:
        - `from datetime import datetime`
        - `now = datetime.now()`
        - `print(now)`
    - Components -
        - print(now.year)
        - print(now.month)
        - print(now.day)
        - print(now.hour)
        - print(now.minute)
- Creating a specific date/time
    - Useful for: deadlines, scheduled jobs, event times
        - `from datetime import datetime`
        - `dt = datetime(2026, 3, 6, 10, 30, 0)`
        - `print(dt)`
- Working with only dates
    - e.g. `from datetime import date`
    - `today = date.today()`
    - `print(today)`
    - e.g. `d = date(2026, 12, 25)`
    - `print(d)`

### Handling durations with `timedelta`
- A timedelta represents a time difference.
- e.g. `from datetime import datetime, timedelta`
    - `now = datetime.now()`
    - `future = now + timedelta(days=7)`
    - `print(future)`
- Add duration
- e.g. `deadline = now + timedelta(hours=2)`
- Subtract
- e.g. `yesterday = now - timedelta(days=1)`

### Calculating time difference
- e.g. `from datetime import datetime`
    - `start = datetime(2026, 3, 6, 10, 0)`
    - `end = datetime(2026, 3, 6, 12, 30)`
    - `duration = end - start`
    - `print(duration)`
    - `print(duration.seconds)`
    - `print(duration.total_seconds())`

### Parsing strings into dates
- Often timestamps come as strings from APIs, logs, or databases.
- e.g. `from datetime import datetime`
    - `date_str = "2026-03-06 10:45:00"`
    - `dt = datetime.strptime(date_str, "%Y-%m-%d %H:%M:%S")`
    - `print(dt)`

| Token | Meaning |
|---|---|
| %Y | Year |
| %m | Month |
| %d | Day |
| %H | Hour |
| %M | Minute |
| %S | Second |

### Converting DateTime --> String
- e.g.`from datetime import datetime`
    - `now = datetime.now()`
    - `formatted = now.strftime("%d-%m-%Y %H:%M")`
    - `print(formatted)`

### Working with Timestamps
- A timestamp = seconds since Jan 1, 1970 (Unix epoch).
- e.g.`from datetime import datetime`
    - `now = datetime.now()`
    - `ts = now.timestamp()`
    - `print(ts)`
- Convert timestamp → datetime:
    - `datetime.fromtimestamp(ts)`

### Handling Deadlines
- Check if deadline passed
- e.g. `from datetime import datetime`
    - `deadline = datetime(2026, 3, 6, 18, 0)`
    - `if datetime.now() > deadline: print("Deadline passed")`
    - `else: print("Still time")`

### Timezone Handling
- By default Python datetime is naive (no timezone). Use timezone-aware datetime.
- e.g. `from datetime import datetime, timezone`
    - `now = datetime.now(timezone.utc)`
    - `print(now)`

### Converting Between Timezones
- Use zoneinfo (Python 3.9+)
- e.g. `from datetime import datetime`
    - `from zoneinfo import ZoneInfo`
    - `utc_time = datetime.now(ZoneInfo("UTC"))`
    - `india_time = utc_time.astimezone(ZoneInfo("Asia/Kolkata"))`
    - `print(india_time)`

### Real World Example: Task Deadline System

- e.g. ` from datetime import datetime, timedelta`
    - `task_created = datetime.now()`
    - `deadline = task_created + timedelta(hours=24)`
    - `remaining = deadline - datetime.now()`
    - `print("Remaining:", remaining)`

### Important Edge Cases
- Naive vs Aware Datetimes
    - e.g. `naive = datetime.now()`
        - `aware = datetime.now(timezone.utc)`
        - `naive > aware`   # ERROR
    - **Always keep both either aware or naive.**

- Daylight Saving Time (DST)
    - e.g. in US - 2:00 AM → 3:00 AM
    - If you add 1 hour during DST transition, results may be unexpected
    - Solution: Use timezone-aware datetimes.

- Leap Year
    - e.g. `date(2024, 2, 29)`  # valid
    - `date(2025, 2, 29)`  # ValueError
    - Always validate user input.

- Month Overflow
    - e.g. This is invalid:
    - `datetime(2026, 13, 1)`
    - Use timedelta instead.

- Negative Timedeltas
    - e.g. `end = datetime(2026, 3, 6)`
    - `start = datetime(2026, 3, 7)`
    - `print(end - start)`

- Microseconds Precision
    - e.g. `datetime.now()`
    - Returns: 2026-03-06 15:40:22.123456
    - Precision can cause issues in comparisons.
    - Often truncate: `dt.replace(microsecond=0)`

### Pro-Level Patterns

- Measuring Execution Time
    - `from datetime import datetime`
    - `start = datetime.now()`
    -  # some task `for i in range(1000000):`
    - `pass`
    - `end = datetime.now()`
    - `print(end - start)`
    - Better alternative: time.perf_counter().

- Expiring Tokens
    - Example: login tokens expire after 30 minutes.
    - `token_created = datetime.now()`
    - `expiry = token_created + timedelta(minutes=30)`
    - `if datetime.now() > expiry: `
    - `print("Token expired")`

- Scheduling Systems
    - e.g. `next_run = datetime.now() + timedelta(hours=6)`
    - Used in: cron jobs, Airflow, distributed systems

- Performance Considerations
    - datetime operations are fast, but repeated conversions can be expensive.
    - Avoid: datetime.strptime()  # inside tight loops
    - Parse once, reuse.

- Best Practices (Pro Tips)

✔ Always store timestamps in UTC
✔ Convert to local timezone only for display
✔ Avoid mixing naive and aware datetimes
✔ Use timedelta instead of manual arithmetic
✔ For heavy date logic use: pandas, pendulum, arrow

### Common Interview Question Example

- Problem: Check if a subscription expired
    - `from datetime import datetime, timedelta`
    - `signup = datetime(2026, 1, 1)`
    - `expiry = signup + timedelta(days=30)`
    - `if datetime.now() > expiry:`
    - `print("Expired")`
