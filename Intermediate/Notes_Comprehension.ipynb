## Comprehension
- Concise syntax patterns for creating collections like lists, dictionaries, and sets in a single line of code.
- List comprehensions, the most common type, allow you to transform and filter data in a clear, expressive syntax.
- Dictionary and set comprehensions follow similar patterns.


### List Comprehension
- **Tool** for creating lists by iterating over an iterable and optionally applying a condition
- Concise, readable code that performs transformations or filtering.
- It is faster to write and faster for the computer to read.
- Syntax: `new_list = [expression for item in iterable]`
    - e.g. `num_list = [x**2 for x in range(10)]`
- Add conditional logic(filtering) to a list comprehension by including an `if` statement within the comprehension
    - e.g. `evens = [i for i in range(20) if i % 2 == 0]`
- *Faster than a `for` loop because it’s optimized for performance by Python’s internal mechanisms*
- Nested loops in List Comprehensions.
- Syntax: `[expression for item in list for item2 in list2 if condition]`. e.g. below:

```python
# Execution goes from left to right and then value is added to the list if the condition is satisfied
evens_pro_tuple = [(i, j) for i in range(20) for j in range(10) if i % 2 == 0]
print(f"Pro Tuple: {evens_pro_tuple}")

# adding multiple if condition to nested loop
evens_pro_tuple_gt_ten = [(i, j) for i in range(20) for j in range(10) if i % 2 == 0 and j % 2 == 0 if i > 10]
print(f"Pro Tuple: {evens_pro_tuple_gt_ten}")
```

### Dictionary Comprehension
- Creates dictionaries using a similar syntax to list comprehensions:
- Syntax: `{key: value for item in list if condition}`
- Just like List Comprehensions, you can add if statements and nested loops to Dictionary Comprehensions.

```python
{x: x**2 for x in range(5)}
```

```python
# Swapping Keys and Values
original = {"A": 1, "B": 2}
swapped = {value ** 2: key for key, value in original.items()}
print(swapped) # {1: "A", 2: "B"}
```

```python
# Intro to zip() method
values = ["Tony", 45, "Ironman"]
keys = ["name", "age", "role"]

# Creating a dict from two lists using zip()
# zip() pairs them up: ("name", "Tony"), ("age", 45)...
# Just like dict.items(), zip also gives us (key, value) pairs from a dict
print(list(zip(keys, values)))

# Dict comprehension to create a dict from two lists
my_dict = {k: v for k, v in zip(keys, values) if isinstance(v, str)}

print(my_dict)
```

### Set Comprehension
- Creates sets using the comprehension syntax with curly braces
- Set Comprehension creates a set, which means:
    - unique elements
    - unordered collection
- Syntax: `{expression for item in iterable}`

```python
emails = ["a@gmail.com", "b@yahoo.com", "c@gmail.com"]

domains = {email.split("@")[1] for email in emails}
print(domains)
```

### Tuple Comprehensions (Generators)
- You can also create tuples with comprehensions, but they are called **Generators**.
- They use parentheses instead of square brackets.
- Syntax: `(expression for item in list)`
- Are more memory efficient than lists because they generate items on the fly instead of storing them all in memory.

```python
# tuple comprehension
numbers = [1, 2, 3, 4, 5]
squares = [num ** 2 for num in numbers] # Memory is allocated
print(squares)

squares_tup = (num ** 2 for num in numbers) # Memory is not allocated
print(squares_tup)
print(next(squares_tup))
print(next(squares_tup))
print(next(squares_tup))
print(next(squares_tup))
print(next(squares_tup))
print(next(squares_tup)) # StopIteration Exception: if we do next on a generator when all values are already consumed
```

```python
# Nested Comprehension: Comprehensions can be nested for more complex operations
[[i+j for j in range(3)] for i in range(3)]
```

#### When not to use Comprehensions
- Warning: If your logic is super complex (more than 2 if statements), do not use a comprehension. Go back to a normal loop. 
Code is for humans to read first, computers second.
