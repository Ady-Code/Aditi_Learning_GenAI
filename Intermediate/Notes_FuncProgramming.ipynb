## Functional Programming
- Tiny, throwaway function for a single task
- Python is a multi-paradigm language that fully supports functional programming concepts, treating functions as first-class citizens.
- While it is not a purely functional language, developers can write **clean, modular, and testable code** by applying its functional features.
- Core Concepts in Python
    - Pure Functions: These functions, central to functional programming, have 2 main properties. One, they always return the same output for the same input and 2nd, they have no side effects (e.g., they do not modify global variables or perform I/O operations).
    - Immutability: In functional programming, we can’t modify a variable after it’s been initialized. We can create new variables – but we can’t modify existing variables.
    - First-Class and Higher-Order Functions: In Python, functions are objects that can be **assigned to variables, passed as arguments, or returned from other functions**, enabling the use of higher-order functions.
    - Recursion: There are no "for" or "while" loop in functional languages. Iteration in functional languages is implemented through recursion.
    - Lambda Functions: Python offers anonymous, single-expression functions using the lambda keyword, which are commonly used with higher-order functions for concise operations.
- *The functional style emphasizes using immutable data (like tuple, str, and frozenset) to prevent unintended state changes, which simplifies debugging and testing.*
- In Python, functions are **first-class citizens**. This means that functions have the same characteristics as values like strings and numbers. Anything you would expect to be able to do with a string or number, you can also do with a function.
    - For example, you can assign a function to a variable. You can then use that variable the same way you would use the function itself.
- Functional Programming concepts:
    - **Lambda Functions**: Anonymous, one-line functions.
    - **Map & Filter**: Tools to process data without writing loops.
    - **Reduce**: Reduces the iterable to a single value.

```python
def func():
    print("I am function func()!")
func()

another_name = func # creates a new reference to func() named another_name
another_name()
```

```python
def inner():
    print("I am function inner()!")
def outer(function):
    function()
outer(inner) # pass a function to another function as an argument
```

### Lambda Functions
- Small, anonymous function
- one-liner code
- Syntax: `lambda arguments : expression`
- e.g.: `lambda s: s[::-1]`

```python
lambda s: s[::-1]
callable(lambda s: s[::-1]) #returns True if the argument passed to it appears to be callable
```

```python
def reverse(s):
    return s[::-1]
reverse("I am a string")
# OR

reverse = lambda s: s[::-1]
reverse("I am a string")
# OR

(lambda s: s[::-1])("I am a string")
```

```python
(lambda x,y,z: (x + y + z) / 3)(9, 9, 9)

(lambda x1, x2, x3: (x1 + x2 + x3) / 3)(1.4, 1.1, 0.5)
```

```python
animals = ["ferret", "vole", "dog", "gecko"]
sorted(animals, key=lambda s: -len(s))
```

```python
print((lambda x: "even" if x % 2 == 0 else "odd")(2))
print((lambda x: "even" if x % 2 == 0 else "odd")(3))
```

```python
# Don't replace every function with a lambda! 
# Use them when you need a short function for a short period of time (usually inside another function).

# (lambda x: x, x ** 2, x ** 3)(3) # NameError: name 'x' is not defined
print((lambda x: (x, x ** 2, x ** 3))(3))

print((lambda x: [x, x ** 2, x ** 3])(3))

print((lambda x: {1: x, 2: x ** 2, 3: x ** 3})(3))

print(f"- {(lambda s: s[::-1])('I am a string')} -") # f-string: lambda expressions are not allowed without parentheses
```

### Map and Filter
#### map()
- With map(), apply a function to every item in a list.
- Syntax: `map(function, list)`
- The map() function will return an **iterator** that yields the results

```python
# map() doesn’t return a list, it returns a map object, which is an iterator
animals = ["cat", "dog", "hedgehog", "gecko"]
map(reverse, animals)

# To obtain the values from the iterator, you need to either iterate over it or use list()
iterator = map(reverse, animals)
# for animal in iterator: # Iterate over it
#     print(animal)
print(list(iterator)) # use list()
```

```python
# use an anonymous lambda function instead of the above
animals = ["cat", "dog", "hedgehog", "gecko"]
iterator = map(lambda s: s[::-1], animals)
list(iterator)

# OR Combining it all into one line
list(map(lambda s: s[::-1], ["cat", "dog", "hedgehog", "gecko"]))
```

```python
# If the iterable contains items that aren’t suitable for the specified function, then Python raises an exception
list(map(lambda s: s[::-1], ["cat", "dog", 3.14159, "gecko"])) # TypeError: 'float' object is not subscriptable
```

```python
# Calling map() with multiple Iterables
def add_three(a, b, c):
    return a + b + c

# list(map(add_three, [1, 2, 3], [10, 20, 30], [100, 200, 300]))

# OR

list(map(lambda a, b, c: a + b + c, 
         [1, 2, 3, 4], [10, 20, 30, 40], [100, 200, 300, 400]))
```

#### filter()
- allows you to select—or filter—items from an iterable based on evaluation of the given function.
- They are faster and cleaner than writing `for` loops.
- Keep only the items where the function returns `True`
- If the function returns anything which evaluates to `False`, value will be filtered out.
- Remember False Values in Python : `False`, `None`, `0`, `0.0`, `0j`, `''`, `[]`, `()`, `{}`, `set()`, `range(0)`

```python
gt_100 = list(filter(lambda x: x > 100, [1, 111, 2, 222, 3, 333]))
print(gt_100)

list(filter(lambda x: x % 2 == 0, range(10)))
```

```python
#  filter a list of animals so that only uppercase values remain
animals = ["cat", "Cat", "CAT", "dog", "Dog", "DOG", "emu", "Emu", "EMU"]
only_upper = list(filter(lambda animal: animal.isupper(), animals))
print(only_upper)
```

```python
# filter() only keeps the non-empty strings which you then use to create a new list
animals_and_empty_strings = ["", "cat", "dog", "", ""]
list(filter(lambda s: s, animals_and_empty_strings))
```

```python
# The Square of Even Numbers

numbers = [1, 3, 10, 45, 6, 50]
list(map(lambda num: num ** 2 ,filter(lambda x: x % 2 == 0, numbers)))
```

- **Note**: `list()` Casting is Mandatory
- In Python 3, `map()` and `filter()` return "Iterators" (objects waiting to run), not Lists.
- Always wrap them: `list(map(...))`.

### reduce()
- Syntax: `reduce(function, list, initial_value)`
- Used to reduce a list into a single value. 
- reduce is not a built-in fucntion. You need to import it from functools module.
    - `from functools import reduce`

```python
from functools import reduce

numbers = [1, 2, 3, 4]

def get_product(x, y):
    return x * y

# in first iteration, x will be 0th element and y will be 1st element from list and return value will be product of the two. 
# in next iteration, x will be result for previous return value and y will be next iterable from the list. 
result = reduce(get_product, numbers)
print(result)
```

```python
result = reduce(lambda x, y: x * y, numbers)
print(result)
```

```python
# example of using initial value in reduce
from functools import reduce

numbers = [1, 2, 3, 4]
result = reduce(lambda x, y: x * y, numbers, 100)
print(result)
```

```python
# find the maximum value in a list
num_list = [23, 49, 6, 32]
reduce(lambda x , y: x if x > y else y,num_list)
```

### Generators & The yield Keyword (Memory Magic)
- In a normal function, when Python hits the return keyword, the function spits out the final result and dies. All its internal variables are erased from memory.
- Instead of return, use the keyword yield. yield means: "Here is one piece of data. Now PAUSE the function, freeze my variables, and wait until you ask for the next piece."
- A generator:
    - produces a value
    - pauses execution
    - resumes when the next value is requested
- produces values one at a time instead of returning everything at once.
- If you just call a generator function, it doesn't run the code inside. 
It creates a "Generator Object" waiting to be triggered. To trigger it and pull the data out, we use the built-in `next()` function, or we loop through it with a `for` loop.

```python
# 1. A Normal Function (Returns a list all at once)
def normal_function():
    return [1, 2, 3]
normal_function()
```

```python
# 2. A Generator Function (Yields one item at a time)
def generator_function():
    print("  -> Generating first number...")
    yield 1
    
    print("  -> Generating second number...")
    yield 2
    
    print("  -> Generating third number...")
    yield 3

generator_function()
```

```python
# The Real Power: Infinite Sequences & Large Data
# Because generators only hold one item in memory at a time, 
# they can generate infinite sequences without ever crashing your computer! 
# This is heavily used in Data Engineering and Web Scraping.
def infinite_counter(start=0):
    current = start
    while True: # An infinite loop that WON'T crash your RAM!
        yield current
        current += 1
counter = infinite_counter(100)
print("Value:", next(counter))
print("Value:", next(counter))
print("Value:", next(counter))
print("Value:", next(counter))
print("Value:", next(counter))
```

```python
# The function remembers where it stopped.
def demo():
    print("Start")
    yield 1
    print("Middle")
    yield 2
    print("End")
g = demo()
print(next(g))
print(next(g))
# print(next(g)) # Python raises a StopIteration error on calling next() when the generator is empty
```
