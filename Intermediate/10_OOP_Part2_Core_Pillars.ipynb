## Topic 1: Composition (Ownership + Lifecycle Control)

Composition means strong ownership. The contained object cannot exist independently.If the parent object is destroyed, the child object also disappears.

````python
# Part 1: The Component
class IDCard:
    def __init__(self,emp_id, emp_name):
        self.__emp_id = emp_id
        self.emp_name = emp_name
    
    def get_id_card(self):
        return f"{self.emp_name} has ID: {self.__emp_id}"
    
    def update_name(self, new_name):
        self.emp_name = new_name
        print(f"Name updated to {self.emp_name}")
    
# Part 2: The Main Object
class Employee:
    def __init__(self, name, department, salary, emp_id):
        self.name = name
        self.department = department
        self.__salary = salary # Private
        # The Developer "Part Of" Laptop
        self.id_card = IDCard(emp_id, self.name)

    def update_name(self, new_name):
        self.name = new_name
        self.id_card.update_name(new_name)
        self.id_card.emp_name = new_name

    def work(self):
        # The Developer delegates the compiling task to their Laptop
        result = self.id_card.get_id_card()
        print(self.id_card._IDCard__emp_id)
        print(f"{result} and works in {self.department}.")

# id_card = IDCard("EMP-123", "Bhaskar")
# id_card.get_id_card()

employee = Employee("Bhaskar", "Engineering", 90000, "EMP-123")
# print(employee.id_card.__emp_id)
# employee.update_name("Bhaskar Chandraul")
employee.id_card.emp_name = "Bhaskar Chandraul"
employee.name = "Bhaskar Chandraul"
print(employee.name)
print(employee.id_card.emp_name)
````

### Constraints in Composition
1. **No access to Private Attribute or Method**: Parent class using the child function will not be able to access private attributes and methods
2. **Name Mangling**: You can access private attribute and methods using name mangling.

## 🧩 Topic 2: Aggregation ("Reference + Independent Lifecycle")
Beginners try to inherit everything. They might say: `class Developer(Laptop):`. 
But wait... A Developer **IS NOT** a Laptop. A Developer **HAS A** Laptop!

When an object is made up of other objects, we use **Aggregation**. We pass the `Laptop` object *into* the `Developer` object.

The contained object can exist independently. In this case Lapton also can exist without Employee.

````python
# Part 1: The Component
class Laptop:
    def __init__(self, brand, memory):
        self.brand = brand 
        self.memory = memory
    
    def compile_code(self):
        return f"Compiling fast on {self.memory}GB RAM!"   
    
    def update_brand(self, new_brand):
        self.brand = new_brand
        print(f"Brand updated to {self.brand}")

# Part 2: The Main Object
class Employee:
    def __init__(self, name, department, salary, laptop_object):
        self.name = name
        self.department = department
        self.__salary = salary # Private
        # The Developer "Part Of" Laptop
        self.laptop = laptop_object
        
    def work(self):
        # The Developer delegates the compiling task to their Laptop
        print(self.laptop.brand)
        return self.laptop.compile_code()

laptop = Laptop("Dell", 8)

employee = Employee("Bhaskar", "Engineering", 90000, laptop)
# employee.work()

employee.laptop.update_brand("Apple")
print(laptop.brand)
print(employee.laptop.brand)

# del employee

````

### Constraints in Aggregation
1. **No access to Private Attribute or Method**: Parent class using the child function will not be able to access private attributes and methods
2. **Name Mangling**: You can access private attribute and methods using name mangling.

### The Key Principle

The relationship is not about the real-world object itself, it is about how your program models the relationship and manages object lifetime.

When deciding between composition and aggregation, ask two questions.

1. Who creates the object?
2. Who controls its lifetime?

Those two answers determine the relationship.

## 🌳 Topic 3: Inheritance ("Is-A")

Inheritance allows a class to reuse properties and methods of another class. Instead of rewriting the same code repeatedly, a new class can extend an existing class.

**Rule of Thumb:** Only use Inheritance if you can say "{Child_Class} IS A {Parent_Class}". Example: Car IS A Vehicle, Manager IS AN Employee, Dog IS AN Animal etc.

### The "Is-A" Relationship

**Tip**: How do you know if you should use Inheritance?

**Test**: Ask "Is a Dog an Animal?" (Yes -> Inherit). "Is a Car a Wheel?" (No -> Don't inherit. A Car HAS a Wheel. That's different).

````python
# The Parent Class 
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.__salary = salary # Private
        
    @property
    def salary(self):
        return self.__salary
        
    def get_details(self):
        return f"{self.name} works in {self.department}."
````

````python
# Creating Child Class without inheriting from Parent Class (No Inheritance)
# The Child Class

class Developer:
    def __init__(self, name, department, salary, programming_language):
        self.name = name
        self.department = department
        self.programming_language = programming_language
        self.__salary = salary

    @property
    def salary(self):
        return self.__salary
    
    def get_details(self):
        return f"{self.name} works in {self.department}. He writes code in {self.programming_language}."
````

Instead of copy-pasting the `name`, `department`, and `__salary` code into a new `Developer` class, we let the `Developer` **inherit** from `Employee`.

## 🔄 Topic 4: Method Overriding

Sometimes, the Child needs to behave differently than the Parent.
If we define a method in the Child with the **same name** as the Parent, the Child's version "wins."
This is called **Overriding**. Method Overloading is a type of polymorphism.

````python
# The Parent Class 
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.__salary = salary # Private
        
    @property
    def salary(self):
        return self.__salary

    def get_salary(self):
        return self.__salary 

    def update_salary(self, new_salary):
        if new_salary < 0:
            raise ValueError("Salary cannot be negative!")
        self.__salary = new_salary   
                
    def get_details(self):
        return f"{self.name} works in {self.department}."

class Developer(Employee):
    def write_code(self):
        return f"{self.name} is typing Python code... 💻"

    def get_details(self):
        return f"{self.name} writes code in Python."
    
developer = Developer("Alice", "Engineering", 90000)
developer.get_details()
````

See how same method `get_detials` has different behaviour for two differnet class. 

````python
emp1 = Employee("Bob", "HR", 60000)
print(emp1.get_details())
dev1 = Developer("Alice", "Engineering", 90000)
print(dev1.get_details())
````

We use `super()` to tell the Parent class to do its normal setup first, and then we add our specific changes

````python
# The Parent Class 
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.__salary = salary # Private
        
    @property
    def salary(self):
        return self.__salary

    def get_salary(self):
        return self.__salary 

    def update_salary(self, new_salary):
        if new_salary < 0:
            raise ValueError("Salary cannot be negative!")
        self.__salary = new_salary   
                
    def get_details(self):
        return f"{self.name} works in {self.department}."

class Developer(Employee):
    def __init__(self, name, department, salary, programming_language):
        self.programming_language = programming_language
        super().__init__(name, department, salary)
        
    def write_code(self):
        return f"{self.name} is typing Python code... 💻"

    def get_details(self):
        return f"{super().get_details()} He writes code in Python."
    
developer = Developer("Alice", "Engineering", 90000, "Python")
developer.get_details()
````

#### Using Method Overloading and Calling Parent's Method

What if we wanted to add more capablilities to Developer's `get_detials` method using the Employee's `get_details` method. 

### Why `super()`?

**The Struggle**: Beginners will try to copy-paste `self.name = name` into the Child's init.

The Fix: *"If the way we store names changes in the Parent class later, your Child class will break. `super()` ensures the Child always follows the Parent's rules."*

### Edge Cases for Super

#### Calling Super outside the class

````python
class Developer(Employee):
    def __init__(self, name, department, salary, programming_language):
        # 1. Let the Parent (Employee) handle the basic setup
        super().__init__(name, department, salary)
        
        # 2. Add the specific Developer attribute
        self.programming_language = programming_language

    def get_details(self):
        # using super to call parent's method
        return super().get_details() + f" They code in {self.programming_language}."    
      
    def write_code(self):
        print(f"{self.name} is typing {self.programming_language} code... 💻")

dev1 = Developer("Alice", "Engineering", 90000, "Python")
print(dev1.super().get_details()) # type: ignore
````

#### Accessing attributes using super()

````python
class Developer(Employee):
    def __init__(self, name, department, salary, programming_language):
        # 1. Let the Parent (Employee) handle the basic setup
        super().__init__(name, department, salary)
        self.name = f"MR. {name}"
        # 2. Add the specific Developer attribute
        self.programming_language = programming_language

    def get_details(self):
        # using super to call parent's method
        return super().get_details() + f" They code in {self.programming_language}."    
      
    def get_department(self):
        return super().department


dev1 = Developer("Alice", "Engineering", 90000, "Python")
print(dev1.get_details())
````

### Single Inheritance
Single inheritance means one child class inherits from one parent class. The child class automatically gets all accessible attributes and methods of the parent.

```bash
Parent
   ↓
Child
```

````python
class Developer(Employee):
    def __init__(self, name, department, salary, programming_language):
        # 1. Let the Parent (Employee) handle the basic setup
        super().__init__(name, department, salary)
        
        # 2. Add the specific Developer attribute
        self.programming_language = programming_language

    def get_details(self):
        # using super to call parent's method
        return super().get_details() + f" They code in {self.programming_language}."    
      
    def write_code(self):
        print(f"{self.name} is typing {self.programming_language} code... 💻")

dev1 = Developer("Alice", "Engineering", 90000, "Python")
print(dev1.get_details())
print(dev1.write_code())
````

### Multilevel Inheritance

Multilevel inheritance means a class inherits from a class that already inherited from another class. Each level can add new behavior. A class can be parent and can be child also. 

```bash
Grandparent
   ↓
Parent
   ↓
Child
```

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    
    def show_employee(self):
        print(f"Employee: {self.name}, Salary: {self.salary}")


class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)
        self.team_size = team_size
    
    def show_manager(self):
        print(f"{self.name} manages {self.team_size} people")


class Director(Manager):
    def __init__(self, name, salary, team_size, department):
        super().__init__(name, salary, team_size)
        self.department = department
    
    def show_director(self):
        print(f"{self.name} heads the {self.department} department")



d = Director("Anita", 250000, 10, "Engineering")

d.show_employee()
d.show_manager()
d.show_director()
````

### Hierarchical Inheritance

Hierarchical inheritance means multiple child classes inherit from the same parent class. Different child classes share common behavior from the same parent.

```bash
        Employee
      /     |    \
Developer Tester Manager
```

````python
class Employee:
    
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    
    def get_details(self):
        print(f"{self.name} works for the company")


class Developer(Employee):
    
    def code(self):
        print(f"{self.name} writes code")


class Tester(Employee):
    
    def test(self):
        print(f"{self.name} tests software")


class Manager(Employee):
    
    def manage(self):
        print(f"{self.name} manages the team")


dev = Developer("Ravi", 120000)
tester = Tester("Sneha", 100000)
mgr = Manager("Arjun", 150000)

dev.get_details()
dev.code()

tester.get_details()
tester.test()

mgr.get_details()
mgr.manage()
````

### Multiple Inheritance(Diamond Problem)
Multiple inheritance means a class inherits from more than one parent class. The child class receives behavior from both parent classes.

```bash 

        Employee
        /     \
   Manager   TechLead
        \     /
      TechManager

```

````python
class Employee:
    def work(self):
        print("Employee works")


class Manager(Employee):
    def manage(self):
        print("Manager manages team")
    
    def work(self):
        print("Manager works")


class TechLead(Employee):
    def design_system(self):
        print("TechLead designs system")
    
    def work(self):
        print("TechLead works")


class TechManager(Manager, TechLead):
    def role(self):
        print("TechManager handles both management and technical leadership")
    
    # def work(self):
    #     return "techmanager works."


tm = TechManager()

tm.work()
tm.manage()
tm.design_system()
tm.role()
````

## 👻 Topic 5: Abstraction (Enforcing Rules)
Imagine you are the CEO. You decree: *"Every employee must have a method to calculate their end-of-year bonus, but the math is different for every role!"*

You can make `Employee` an **Abstract Base Class (ABC)**. 
It forces all Child classes (Developer, Manager) to write a `calculate_bonus()` method. If they don't, Python crashes.

````python
from abc import ABC, abstractmethod

# 1. Inherit from ABC to make it abstract
class AbstractEmployee(ABC):
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.__salary = salary
        
    @property
    def salary(self):
        return self.__salary
    
    @salary.setter
    def salary(self, new_salary):
        if new_salary < 0:
            raise ValueError("Salary cannot be negative!")
        self.__salary = new_salary
        
    # 2. Use the decorator to create a strict rule
    @abstractmethod
    def get_details(self):
        pass # No logic here, just the rule!
    
# 3. The Child MUST follow the rule
class Developer(AbstractEmployee):
    def __init__(self, name, salary, department, programming_language):
        super().__init__(name, salary, department)
        self.programming_language = programming_language
    # If we delete this method, Python will crash!
    def get_details(self):
        return f"{self.name} is interning in {self.department} and uses {self.programming_language}."

    def calculate_bonus(self):
        return 500 # Interns get a flat $500 bonus
````

````python
emp = AbstractEmployee("Ghost", 0, "Engineering") # 🛑 CRASH! Cannot hire an abstract concept.
````

````python
timmy = Developer("Timmy", 30000, "Engineering", "Python")
````

````python
print(f"{timmy.name}'s Bonus: ${timmy.calculate_bonus()}")
print(timmy.get_details())
````

## Topic 5: Polymorphism in Action

In Python, Polymorphism allows different classes to be treated as instances of the same general class through the same interface. The word literally means "many forms"—the same method name (interface) takes on different behaviors depending on which object is calling it.

The best way to see this is through a Payment System. Whether you pay via Credit Card or PayPal, the "action" is the same (`process_payment`), but the internal logic is very different.

````python
class CreditCard:
    def process_payment(self, amount):
        return f"Processing credit card payment of ${amount}."
class PayPal:
    def process_payment(self, amount):
        return f"Processing PayPal payment of ${amount}."
class Crypto:
    def process_payment(self, amount):
        return f"Processing Bitcoin payment of ${amount}."
````

````python
class Cat:
    def speak(self): return "Meow"

class Dog:
    def speak(self): return "Woof"

class Robot:
    def speak(self): return "Beep Boop"

# This function is polymorphic but uses NO inheritance/overriding
def make_it_talk(entity):
    print(entity.speak())

make_it_talk(Cat())
make_it_talk(Robot())
make_it_talk(Dog())
````

````python
# The Parent Class (From Day 08)
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.__salary = salary # Private
        
    @property
    def salary(self):
        return self.__salary
        
    def get_details(self):
        return f"{self.name} works in {self.department}."

# The Child Class
# Syntax: class Child(Parent):
class Developer(Employee):
    def __init__(self, name, department, salary, programming_language):
        # 1. Let the Parent (Employee) handle the basic setup
        super().__init__(name, department, salary)
        
        # 2. Add the specific Developer attribute
        self.programming_language = programming_language

    def get_details(self):
        return super().get_details() + f" They code in {self.programming_language}."  
      
    def write_code(self):
        print(f"{self.name} is typing {self.programming_language} code... 💻")
````

See how same method `get_detials` has different behaviour for two differnet class. 


````python
emp1 = Employee("Bob", "HR", 60000)
print(emp1.get_details())
dev1 = Developer("Alice", "Engineering", 90000, "Python")
print(dev1.get_details())
````

````python
class Payment:
    def process_payment(self, amount):
        raise NotImplementedError("Subclass must implement abstract method")

class CreditCard(Payment):
    def process_payment(self, amount):
        return f"Processing credit card payment of ${amount} (Applying 2% transaction fee)."

class PayPal(Payment):
    def process_payment(self, amount):
        return f"Processing PayPal payment of ${amount} (Redirecting to secure portal)."

class Crypto(Payment):
    def process_payment(self, amount):
        return f"Processing Bitcoin payment of ${amount} (Waiting for blockchain confirmation)."
````

````python
# --- THE POLYMORPHIC FUNCTION ---
def checkout(payment_method, amount):
    # This function doesn't care what 'type' payment_method is.
    # It only cares that it HAS a .process_payment() method.
    print(payment_method.process_payment(amount))

# Using different objects in the same function
methods = [CreditCard(), PayPal(), Crypto()]

for method in methods:
    checkout(method, 100)
````

### Why is this powerful?
**Flexibility**: You can add a new payment method (like ApplePay) next week. You only need to create the class and the process_payment method. You don't have to change the checkout function at all.

**Decoupling**: The checkout logic is separated from the specific details of how each payment works.

**Duck Typing**: Python follows the "Duck Typing" philosophy: "If it walks like a duck and quacks like a duck, it’s a duck." In our example, if an object has a process_payment method, Python treats it as a valid payment method.
