### Most used commands-

| Commands | Windows | Unix |
|---|---|---|
| List files | `ls` or `dir` | |
| Show current folder | `cwd` | |
| Change directory | `cd <folder>` | |
| Go up one level | `cd ..` | |
| View file contents |  `cat filename` or `filename`| |
| Home directory| `cd ~` | |
| Make folder | `mkdir <folder>` | `mkdir <folder>` |
| Create file | `New-Item <filename>` | touch file.txt |
| Delete file | `rm file.txt` | Remove-Item file.txt |

| Move/Rename | `mv <source> <destination>` | Move-Item |
| Copy file | `cp <source> <destination>` | Copy-Item |

- In Windows Command Prompt (CMD), the command to move a file to a new location while simultaneously renaming it is `mv <source> <destination>`
    - e.g. mv file.txt fileHandling --> moves the file 'file.txt' to 'fileHandling'folder with same name
    - e.g. mv file.txt fileHandling\fileNew.txt --> moves the file  'file.txt' to 'fileHandling'folder with a new name
    - e.g. ren file.txt newfile.txt --> renames the file 'file.txt' to 'newfile.txt' in the same directory
