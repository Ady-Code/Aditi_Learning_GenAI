## Python Dictionaries
- Allows you to store key-value pairs for efficient data retrieval and manipulation.
- A dictionary in Python is a **mutable collection of key-value pairs** that allows for efficient data retrieval using unique keys.
- Python’s dictionaries have the following characteristics:
	- Mutable: The dictionary values can be updated in place.
	- Dynamic: Dictionaries can grow and shrink as needed.
	- Efficient: They’re implemented as hash tables, which allows for fast key lookup.
	- Ordered: Starting with Python 3.7, dictionaries keep their items in the same order they were inserted.

- The **keys of a dictionary** have a couple of restrictions. They need to be:
	- Hashable: This means that you can’t use unhashable objects like lists as dictionary keys.
	- Unique: This means that your dictionaries won’t have duplicate keys.
- Whereas the **values in a dictionary** aren’t restricted. They can be of any Python type, including other dictionaries, which makes it possible to have nested dictionaries.
- **Note**- Since dictionaries are collections of pairs, you can’t insert a key without its corresponding value or vice versa.
- **Note**- In some situations, you may want to add keys to a dictionary without deciding what the associated value should be. In those cases, you can use the .setdefault() method to create keys with a default or placeholder value.

````python
config = { "color": "green", "width": 42, "height": 100, "font": "Courier"}
# Access a value through its key
print(config["color"])

# Update a value
config["font"] = "Helvetica"
print(config)
````

### Creating Dictionaries in Python
#### Dictionary Literals
- Define a dictionary by enclosing a comma-separated series of key-value pairs in curly braces({}).
``` {<key_1>: <value_1>,
    <key_2>: <value_2>,
      ...,
    <key_N>: <value_N>,
}
```
- The keys must be **hashable objects** like numbers, strings, or tuples.
```
{42: "aaa", 2.78: "bbb", True: "ccc"}
```
- **In Python, the built-in immutable data types are hashable, and the mutable types are unhashable.**
- You can even use objects like data types and functions as keys:
e.g. below

````python
types = {int: 1, float: 2, bool: 3}
print(types)
# {<class 'int'>: 1, <class 'float'>: 2, <class 'bool'>: 3}
print(types[float])
print(types[bool])
````

````python
# You can use tuples as keys because tuples are immutable: 
a_dict = {(1, 1): "a", (1, 2): "b", (2, 1): "c", (2, 2): "d"}

print(a_dict[(1, 1)]) 
# output: a
print(a_dict[(2, 1)])
# output: c

# You can’t use a tuple that contains mutable objects as a dictionary key:
my_dict = {(1, [1, 1]): "a"}
print(my_dict)
# Output:
# TypeError: cannot use 'tuple' as a dict key (unhashable type: 'list')
````

````python
# If you specify a key a second time during the creation of a dictionary, 
# the second occurrence will override the first:

MLB_teams = {
"Colorado": "Rockies",
"Chicago": "White Sox",
"Chicago": "Cubs", # This will override the previous one
"Boston": "Red Sox",
"Minnesota": "Twins",
"Milwaukee": "Brewers",
"Seattle": "Mariners"}

print(MLB_teams)
````

- Unlike dictionary keys, a dictionary value can be any type of object, including mutable types like lists and dictionaries, as well as user-defined objects:
```
{
"colors": ["red", "green", "blue"],
"plugins": {"py_code", "dev_sugar", "fasting_py"},
"timeout": 3,
"position": Point(42, 21)
}
```
- No restriction against a particular value appearing in a dictionary multiple times
- A **cool way to create dictionaries** from sequences of values is to combine them with the built-in `zip()` function and then call dict() as shown below:
```
places = [
"Colorado",
"Chicago",
 "Boston",
 "Minnesota"]

teams = [
 "Rockies",
 "White Sox",
"Red Sox",
"Twins"]
```
`dict(zip(places, teams))`
- The zip() function takes one or more iterables as arguments and yields tuples that combine items from each iterable.**Order is essential**

#### Using the .fromkeys() Class Method
- The dict data type has a class method called .fromkeys() that lets you create new dictionaries from an iterable of keys and a default value.
`.fromkeys(iterable, value=None, /)`
- The iterable argument provides the keys that you want to include in your dictionary. Even though the input iterable can have duplicate items, the final dictionary will have unique keys as usual.
- The `value` argument allows you to define an appropriate default value for all the keys. This argument defaults to None, which can serve as a good default value in several scenarios.
- e.g. 
```
inventory = dict.fromkeys(["apple", "orange", "banana", "mango"], 0)
print(inventory)
Output: {'apple': 0, 'orange': 0, 'banana': 0, 'mango': 0}
```

### Accessing Dictionary Values
- To retrieve a value from a dictionary, you can specify its corresponding key in square brackets ([]) after the dictionary name: `MLB_teams["Minnesota"]`
- If you refer to a key that isn’t in the dictionary, then Python raises an **exception**:
```
MLB_teams["Indianapolis"]
Traceback (most recent call last):
    ...
KeyError: 'Indianapolis'
```

### Populating Dictionaries Incrementally
- 3 common techniques:
1. Assign keys manually
2. Add keys in a for loop
3. Build a dictionary with a comprehension

#### Assign keys manually

```
person = {}

person["first_name"] = "John"
person["last_name"] = "Doe"
person["age"] = 35

print(person)
Output:
person
{
    'first_name': 'John',
    'last_name': 'Doe',
    'age': 35
}
```

#### Add keys in a for loop
```
squares = {}

for integer in range(1, 10):
    squares[integer] = integer**2

print(squares)
Output:
{1: 1, 2: 4, 3: 9, 4: 16, 5: 25, 6: 36, 7: 49, 8: 64, 9: 81}
```

#### Building Dictionaries With Comprehensions
- Another great tool for creating and populating dictionaries with concise syntax.
```
squares = {integer: integer**2 for integer in range(1, 10)}
print(squares)
Output:
{1: 1, 2: 4, 3: 9, 4: 16, 5: 25, 6: 36, 7: 49, 8: 64, 9: 81}
```

### Retrieving Data From Dictionaries
#### Getting Individual keys: 
- .get(key, default=None)
- Convenient way to retrieve the value associated with a key without checking whether the key exists beforehand.
- First argument is the key.
- 2nd argument, which is optional,  is a default value that will be used if the target key doesn’t exist in the dictionary.
- e.g.

````python
inventory = {"apple": 100, "orange": 80, "banana": 100}

print(inventory.get("apple"))
print(inventory.get("apple",250)) # already exists, so doesn't take the default value
print(inventory.get("mango"))
print(inventory.get("mango",50)) # doesn't exist, so it takes the default value
````

#### Retrieving All the Values
- .values()
- The .values() method returns a dictionary view object, which provides a dynamic view of the values in a dictionary

````python
inventory = {"apple": 100, "orange": 80, "banana": 1}
# Note that any duplicate values will be returned as many times as they occur.
print(inventory.values())
````

#### Accessing All the Keys
- .keys()
- The .keys() method returns a dictionary view object with a dynamic view of the keys in the target dictionary.

````python
inventory = {"apple": 100, "orange": 80, "banana": 100}
# Since dictionary keys are unique, you won’t get any duplicate keys.
print(inventory.keys())
````

#### Getting All the Items or Key-Value Pairs
- .items()
- The .items() method returns a dictionary view containing tuples of keys and values.
- The first item in each tuple is the key, while the second item is the associated value.

````python
inventory = {"apple": 100, "orange": 80, "banana": 100}
print(inventory.items())
````

### Adding Key-Value Pairs and Updating Dictionaries
- .setdefault() and .update() methods
#### Setting One Key
- The .setdefault() method lets you set default values to keys.
- If key is in the dictionary, then the method returns the associated value.
- If key isn’t in the dictionary, it’s inserted with default as its associated value.

````python
inventory = {"apple": 100, "orange": 80}

# When you call .setdefault() with an existing key, you get the associated value.
print(inventory.setdefault("apple")) 

# If the key is missing, you get None—which is the default value—and a new key-value pair is inserted.
inventory.setdefault("mango")
print(inventory)

# If the key is missing and you provide a custom default value, then you get the custom default and a new key-value pair.
inventory.setdefault("banana", 0)
print(inventory)
````

#### Updating a Dictionary
- .update([other])
- The .update() method merges a dictionary with another dictionary or with an iterable of key-value pairs.
- If `other_dict` is a dictionary, then `a_dict.update(other)` merges the entries from `other_dict` into `a_dict`.
- For each key in `other_dict`, you can have one of the following results:
    - If the key isn’t present in `a_dict`, then the key-value pair from `other_dict` is added to `a_dict`.
    - If the key is present in `a_dict`, then the corresponding value in `a_dict` is updated to the value in `other_dict`.

````python
config = {
"color": "green",
"width": 42,
"height": 100,
"font": "Courier",
}

user_config = {
"path": "/home",
"color": "red",
"font": "Arial",
"position": (200, 100),
}

config.update(user_config)
print(config)

config.update([("width", 200), ("api_key", 1234)])
print(config)
````

````python
# Finally, you can also call .update() with keyword arguments

config.update(color="yellow", script="__main__.py")
print(config)
````

### Removing Data From Dictionaries

#### Removing Keys
- .pop(key[, default])
- The .pop() method removes key-value pairs by keys.
- If the key exists, then the method returns its associated value.
- If the key doesn’t exist and default isn’t provided, then you get a KeyError. Otherwise, you get the default value.

````python
inventory = {"apple": 100, "orange": 80, "banana": 100}
inventory.pop("apple")
print(inventory)

inventory.pop("mango")
# KeyError: 'mango'

inventory.pop("mango", 0) # returns the value, used to both delete an item and retain its value

del inventory["banana"] # just removes the item
print(inventory)
````

#### Deleting Items
- .popitem()
- Removes a key-value pair from a dictionary
- This method returns the removed pair as a tuple of the form (key, value).
- The pairs are removed in LIFO (last-in, first-out) order

````python
inventory = {"apple": 100, "orange": 80, "banana": 100}

# items are removed from right to left, starting with the last item added
inventory.popitem() # returns the value
# Output: ('banana', 100)
print(inventory)

inventory.popitem()
# Output: ('orange', 80)
print(inventory)

inventory.popitem()
# Output: ('apple', 100)
print(inventory)

# If the dictionary is empty, then .popitem() raises a KeyError exception:
inventory.popitem() # no further items to remove
# Output: KeyError: 'popitem(): dictionary is empty'
````

#### Clearing Dictionaries
- .clear()
- The .clear() method removes all the items from a dictionary

````python
inventory = {"apple": 100, "orange": 80, "banana": 100}
inventory.clear()
print(inventory)
````
