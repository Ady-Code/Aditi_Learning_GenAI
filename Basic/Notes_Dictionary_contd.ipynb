#### Using operators in Dictionary

```python
# in and not in
MLB_teams = { "Colorado": "Rockies",
"Chicago": "White Sox",
"Boston": "Red Sox",
"Minnesota": "Twins",
"Milwaukee": "Brewers",
"Seattle": "Mariners"}

print("Milwaukee" in MLB_teams) #True; checks the keys
print("Indianapolis" in MLB_teams) #False; checks the keys
print("Indianapolis" not in MLB_teams) #True; checks the keys
print("Milwaukee" in MLB_teams.keys()) #True
print("Indianapolis" in MLB_teams.keys()) #False
print("Indianapolis" not in MLB_teams.keys()) #True
print("Rockies" in MLB_teams.values()) #True
print("Rockies" not in MLB_teams.values()) #False
# you may want to know whether a key-value pair is in the target dictionary
print(("Boston", "Red Sox") in MLB_teams.items()) #True
print(("Boston", "Red Sox") not in MLB_teams.items()) #False
```

```python
# Equality and Inequality: == and !=
print([1, 2, 3] == [3, 2, 1]) # in case of list, the result depends on both the content and the order
print({1: 1, 2: 2, 3: 3} == {3: 3, 2: 2, 1: 1}) #when you compare two dictionaries, the order of those pairs isn’t considered
print({1: 1, 2: 2, 3: 3} != {3: 3, 2: 2, 1: 1})
```

```python
# Union and Augmented Union: | and |=
default_config = {
    "color": "green",
    "width": 42,
    "height": 100,
    "font": "Courier",
}

user_config = {
    "path": "/home",
    "color": "red",
    "font": "Arial",
    "position": (200, 100),
}

config = default_config | user_config # Returns a new dictionary
print(config)
default_config |= user_config # Updates original dictionary
print(default_config)
```

#### Built-in Functions With Dictionaries

```python
# all(): Returns True if all the items in an iterable are truthy and False otherwise
inventory = {"apple": 100, "orange": 80, "banana": 100, "mango": 200}
print(all(inventory.values()))

# Update the stock
inventory["mango"] = 0
print(all(inventory.values()))
```

#### Finding Minimum and Maximum Values: min() and max()

```python
computer_parts = {
"CPU": 299.99,
"Motherboard": 149.99,
"RAM": 89.99,
"GPU": 499.99,
"SSD": 129.99,
"Power Supply": 79.99,
"Case": 99.99,
"Cooling System": 59.99,
}

print(min(computer_parts.values()))
print(max(computer_parts.values()))
```

```python
# Sorting Dictionaries by Keys, Values, and Items: sorted()
students = {
"Alice": 89.5,
"Bob": 76.0,
"Charlie": 92.3,
"Diana": 84.7,
"Ethan": 88.9,
"Fiona": 95.6,
"George": 73.4,
"Hannah": 81.2}

dict(sorted(students.items(), key=lambda item: item[1]))
```

```python
dict(sorted(students.items(), key=lambda item: item[1], reverse=True))
```

```python
# sort the dictionary by its keys
dict(sorted(students.items(), key=lambda item: item[0]))
```

```python
# you can also use sorted() to sort the keys and values
sorted(students)
```

```python
sorted(students.values())
```

```python
# Summing Dictionary Values: sum()
daily_sales = {
"Monday": 1500,
"Tuesday": 1750,
"Wednesday": 1600,
"Thursday": 1800,
"Friday": 2000,
"Saturday": 2200,
"Sunday": 2100 }
sum(daily_sales.values()) / len(daily_sales)
```

#### Traversing Dictionaries by Keys
- There are two different ways you can iterate over the keys of a dictionary. You can either use the dictionary directly, or use the .keys() method.

```python
students = {
"Alice": 89.5,
"Bob": 76.0,
"Charlie": 92.3,
"Diana": 84.7,
"Ethan": 88.9,
"Fiona": 95.6,
"George": 73.4,
"Hannah": 81.2 }

for student in students:
    print(student)
```

```python
for student in students.keys():
    print(student)
```

```python
#  In both above loops, you can access the dictionary values as well:
for student in students:
    print(student, "->", students[student])
```

#### Iterating Over Dictionary Values

```python
MLB_teams = {
"Colorado": "Rockies",
"Chicago": "White Sox",
"Boston": "Red Sox",
"Minnesota": "Twins",
"Milwaukee": "Brewers",
"Seattle": "Mariners"}
for team in MLB_teams.values():
    print(team)
```

#### Looping Through Dictionary Items: .items() method

```python
for place, team in MLB_teams.items():
    print(place, "->", team)
```
