## 🔁 Topic 4: Looping through Dictionaries
You can loop through Keys, Values, or Both.

```python
prices = {"Apple": 2, "Banana": 1, "Cherry": 5}

# Loop through Keys (Default)
print("--- Fruits ---")
for fruit in prices:
    print(fruit)

# Loop through Values
print("--- Prices ---")
for cost in prices.values():
    print(cost)

# Loop through Both (Most useful!)
print("--- Menu ---")
for fruit, cost in prices.items():
    print(f"The {fruit} costs ${cost}")
```

```python
users = {
    "user1": {
        "name": "Alice",
        "age": 25
    },
    "user2": {
        "name": "Bob",
        "age": 30
    }
}

# How to access Bob's age?
# Go to 'user2', then go to 'age'
print(users["user2"]["age"])
```

### Level 3: The Translator 🗣️

2. Ask the user for a word (e.g., "hello").
3. Print the translation.
4. *Bonus:* Use `.get()` so it says "None" if the word isn't found.

```python
# Write your code here for Level 3
spanish = {"hello": "hola", "thank you": "gracias", "yes": "si"}

def translate():
    try:
        word = input("Enter a word")
        print(spanish.get(word))

    except Exception as e:
        print(f"Error: {e}")

translate()
```
