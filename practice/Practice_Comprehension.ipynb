```python
# You have a list of numbers. Create a new list where each number is converted to a string. Store output in variable named `output`.
numbers = [1, 2, 3, 4, 5]
output = None

# Write Your Code Here
output = [str(num) for num in numbers]
print(output)

expected_output = ["1", "2", "3", "4", "5"]
if output == expected_output:
    print("Success!")
```

```python
# You have a list of words. Create a list containing the length of each word. Store output in variable named `output`.
words = ["python", "data", "science", "ai"]
output = None

# Write Your Code Here
output = [len(word) for word in words]
print(output)

expected_output = [6, 4, 7, 2]
if output == expected_output:
    print("Success!")
```

```python
# You have usernames. Create a new list where each username has "user_" added in front. Store output in variable named `output`.
users = ["amit", "neha", "raj", "sara"]
output = None

# Write Your Code Here
output = ["user_" + user for user in users]
print(output)

expected_output = ["user_amit", "user_neha", "user_raj", "user_sara"]
if output == expected_output:
    print("Success!")
```

```python
# You have a list of product dictionaries. Create a list containing only the prices. Store output in variable named `output`.
products = [
    {"name": "Laptop", "price": 50000},
    {"name": "Mouse", "price": 500},
    {"name": "Keyboard", "price": 1500}
]
output = None

# Write Your Code Here
# for product in products:
#     print(product["price"])
output = [product["price"] for product in products]
print(output)

expected_output = [50000, 500, 1500]
if output == expected_output:
    print("Success!")
```

```python
# You have a list of product dictionaries. Create a list containing only the prices. Store output in variable named `output`.
names = ["amit", "neha", "raj", "sara"]
output = None

# Write Your Code Here
output = [name.upper() for name in names]
print(output)

expected_output = ["AMIT", "NEHA", "RAJ", "SARA"]
if output == expected_output:
    print("Success!")
```
