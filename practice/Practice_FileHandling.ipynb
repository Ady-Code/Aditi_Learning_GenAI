```python
import csv

data = [
    ["Name", "Age", "City"],
    ["Aman", 28, "Pune"],
    ["Poonam", 24, "Jaipur"],
    ["Bobby", 32, "Delhi"]
]

csv_file_path = 'output_list.csv'

with open(csv_file_path, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data) # writes all rows at once

print(f"CSV file '{csv_file_path}' created successfully.")
```

```python
import csv

data_dict = [
    {'Name': 'Rajat', 'Age': 30, 'City': 'New York'},
    {'Name': 'Tarun', 'Age': 25, 'City': 'Delhi'},
    {'Name': 'Bobby', 'Age': 35, 'City': 'Pune'}
]

csv_file_path = 'output_dict.csv'
fieldnames = ['Name', 'Age', 'City'] # Defines column order

with open(csv_file_path, mode='w', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader() # Writes the header row
    writer.writerows(data_dict) # Writes all dictionary rows

print(f"CSV file '{csv_file_path}' created successfully.")
```
