```python
# Convert below into lambda function

# def multiply_by_10(x):
#     return x * 10
multiply_by_10 = lambda x : x * 10
print(multiply_by_10(10))

# def is_adult(age):
#     return age >= 18
is_adult = lambda age : age >= 18
print(f"This person is an adult: {is_adult(19)}")

# def get_username(email):
#     return email.split("@")[0]
get_username = lambda email: email.split("@")[0]
print(f"The username is {get_username('abcxyz@=yahoo.com')}")

# def rectangle_area(length, width):
#     return length * width
rectangle_area = lambda length, width : length * width
print(rectangle_area(5,10))

# def calculate_total(price, tax):
#     total = price + tax
#     print("Total calculated")
#     return total
calculate_total = lambda price, tax : price + tax
print(calculate_total(100.0, 99.9))
```

```python
# identify palindrome words
def is_palindrome(word):
    reversed_word = word[::-1]
    return word == reversed_word
words = ("filter", "Ana", "hello", "world", "madam", "racecar")
list(filter(is_palindrome,words))
```

```python
# validate identifiers
words = ["variable", "file#", "header", "_non_public", "123Class"]
list(filter(str.isidentifier, words))
```

```python
# Extracting Even Numbers
numbers = [1, 3, 10, 45, 6, 50]
even_num = list(filter(lambda num: num % 2 ==0 , numbers))
print(even_num)
```

```python
# factorial function using reduce()

# def factorial(x):
#     final = x
#     for i in range(x,2,-1):
#         final *= (i-1)
# factorial(5)
# ----------------------------------------------------
# def factorial(x):
#     final = 1
#     for i in range(1,x+1):
#         final *= i
#     print(final)
# factorial(5)

from functools import reduce
def factorial(n):
    return reduce(lambda x,y : x * y ,range(1, n + 1))
factorial(5)
```

```python
# find the maximum value in a list
num_list = [23, 49, 6, 32]
reduce(lambda x , y: x if x > y else y,num_list)
```

```python
# 2. A Generator Function (Yields one item at a time)
def generator_function():
    print("  -> Generating first number...")
    yield 1
    
    print("  -> Generating second number...")
    yield 2
    
    print("  -> Generating third number...")
    yield 3

my_gen = generator_function()
# Trigger it one step at a time using next()
# print("Value:", next(my_gen))  # Runs up to the first yield, then PAUSES
# print("Value:", next(my_gen)) # Resumes, runs up to the second yield, PAUSES
# print("Value:", next(my_gen))
# print("Value:", next(my_gen)) # Error- StopIteration
print("--- Starting Loop ---")
for number in my_gen:
    print("Value:", number) # Prints 3, then stops safely when the generator is empty
```
