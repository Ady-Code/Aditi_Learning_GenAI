```python
# 2. WITH Type Hinting (Professional)
def calculate_total_pro(price: float, tax: float) -> float:
    """Calculates the total price including tax."""
    return price + (price * tax)

print(calculate_total_pro(100.0, 0.05))
```

```python
# Complex types require the 'typing' module in older Python versions, 
# but in Python 3.9+, you can just use list, dict, etc.
def process_users(users: list[str]) -> dict[str, bool]:
    return {user: True for user in users}

print(process_users(["Alice", "Bob"]))
```

```python
# 1. Assigning a function to a variable (NO parentheses!)
def lower_name(name:str) -> str:
    return name.lower()

all_lower = lower_name
print(all_lower("PYTHON"))
```

```python
# 2. Calling a function inside a function
def lower_char(name:str) -> str:
    return lower_name(name).upper()

lower_char("hello!")
```

```python
# 3. Passing a function into another function
def greet_in_lower(lower_char,name: str) -> str:
    greeting = lower_char(name)
    print(greeting)

greet_in_lower(lower_char,"aditi")
```

```python
def parent():
    print("Printing from parent()")

    def first_child():
        print("Printing from first_child()")

    def second_child():
        print("Printing from second_child()")

    second_child()
    first_child()

parent()
```

```python
def parent(num):
    def first_child():
        return "Hi, I'm Elias"

    def second_child():
        return "Call me Ester"

    if num == 1:
        return first_child
    else:
        return second_child

first = parent(1)
second = parent(2)

first()
second()
```

```python
# 1. Build the Decorator (The Protective Case)
def star_decorator(func):
    def wrapper():
        print("********************") # Do something BEFORE
        func()                      # Run the original function
        print("********************") # Do something AFTER
    return wrapper                  # Return the upgraded function

# 2. Define the function we want to add decorators to
def greet():
    print("Hello, World!")

# 3. Using decorator as a function
greet_decorated = star_decorator(greet)
greet_decorated()
```

```python
def play_game() -> None:
    print("Hi, Would you like to play a game?")

play_game_decorated = star_decorator(play_game)
play_game_decorated()
```

```python
# 2. Apply it using '@'
@star_decorator
def greet():
    print("Hello, World!")

# 3. Call the function normally!
greet()
```

```python
@star_decorator
def play_game() -> None:
    print("Hi, Would you like to play a game?")

play_game()
```

```python
# In-Session Practice 2: The Bracket Decorator
#  Write your decorator here:
def add_brackets(func):
    def wrapper():
        print("[", end="")
        func()
        print("]")
    return wrapper


@add_brackets
def print_name():
    print(" Tony ", end="")

print_name()
# final result : "[ Tony ]"
```

```python
from datetime import datetime

def not_during_the_night(func):
    def wrapper():
        if 7 <= datetime.now().hour < 22:
            func()
        else:
            pass  # Hush, the neighbors are asleep
    return wrapper

def say_whee():
    print("Whee!")

say_whee = not_during_the_night(say_whee)
say_whee()
```

```python
def do_twice(func):
    def wrapper_do_twice(*args, **kwargs):
        func(*args, **kwargs)
        func(*args, **kwargs)
    return wrapper_do_twice

@do_twice
def say_whee(name):
    print(f"Hello {name}")
say_whee(name = "World")
```

```python
def do_twice(func):
    def wrapper_do_twice(*args, **kwargs):
        func(*args, **kwargs)
        func(*args, **kwargs)
        return func(*args, **kwargs)
    return wrapper_do_twice

@do_twice
def say_whee(name):
    print(f"Hello {name}")
    return f"Hi {name}"
say_whee(name = "World")
```
