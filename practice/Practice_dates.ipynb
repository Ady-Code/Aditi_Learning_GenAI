```python
from datetime import datetime

start = datetime(2026, 3, 6, 10, 0)
end = datetime(2026, 3, 6, 12, 30)

duration = end - start

print(start)
print(end)
print(duration)
print(duration.seconds)
print(duration.total_seconds())
```

```python
from datetime import datetime, timedelta

now = datetime.now()

future = now + timedelta(days=7)
print(future)
deadline = now + timedelta(hours=2)
print(deadline)
yesterday = now - timedelta(days=1)
print(yesterday)
```

```python
from datetime import datetime

date_str = "2026-03-06 10:45:00"
dt = datetime.strptime(date_str, "%Y-%m-%d %H:%M:%S")
print(isinstance(dt,datetime))
print(dt)
```

```python
from datetime import datetime as dt
now = datetime.now()
ts = now.timestamp() # seconds since Jan 1, 1970
formatted = now.strftime("%d-%m-%Y %H:%M")
print(formatted)
print(ts)
print(dt.fromtimestamp(ts))
print(isinstance(formatted,datetime))
```

```python
from datetime import datetime
from zoneinfo import ZoneInfo
utc_time = datetime.now(ZoneInfo("UTC"))
india_time = utc_time.astimezone(ZoneInfo("Asia/Kolkata"))
print(india_time)
```

```python
from datetime import datetime, timedelta
task_created = datetime.now()
print(task_created.replace(microsecond=0))
deadline = task_created + timedelta(hours=24)
print(deadline.replace(microsecond=0))
remaining = deadline.replace(microsecond=0) - datetime.now().replace(microsecond=0)
print("Remaining:", remaining)
```
